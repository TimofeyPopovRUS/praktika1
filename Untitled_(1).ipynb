## В рамках бизнес-задачи постройте несколько сводных таблиц

```python

dataframe.pivot_table(values=['mileage', 'mpg'],
                               index='transmission',
                               aggfunc='mean')


```

Эта таблица показывает средний пробег и средний расход топлива автомобиля в зависимости от типа трансмиссии

```python

dataframe.pivot_table(values=['price', 'enginesize'],
                               index='year',
                               aggfunc='mean')


```

Эта таблица показывает среднюю цену и средний размер двигателя по году производства автомобиля

```python


dataframe.pivot_table(values='price',
                               index='transmission',
                               aggfunc=['max', 'min'])

```

Эта таблица показывает максимальную и минимальную цену автомобиля в зависимости от типа трансмиссии

```python
dataframe.groupby('year').agg({
    'price': ['mean', 'min', 'max'],
    'mileage': ['mean', 'min', 'max'],
    'tax': ['mean', 'min', 'max'],
    'mpg': ['mean', 'min', 'max'],
    'enginesize': ['mean', 'min', 'max'] })


```

Сгруппировали данные по году выпуска и вычислили среднее, минимальное и максимальное значения для различных показателей стоимости автомобиля

```python

sns.scatterplot(x='year', y='price', hue='transmission', data=dataframe)
plt.title('Зависимость цены от года производства')
plt.show()
```

На данном графике изображена зависимость между годом производства автомобиля и его ценой

Можно сделать некоторые выводы: механическая коробка передач распространена среди более старых автомобилей и их цена ниже посравнению с автомобилями с автоматами того же года, независимо от типа трансмиссии, ценовые характеристики автомобилей становятся более равномерно распределёнными с течением времени, что может говорить о выравнивании технологий и стоимости их внедрения








```python
plt.figure(figsize=(8, 5))
plt.hist(dataframe['price'], bins=10, color='orange', edgecolor='black')
plt.title('Распределение кол-ва автомобилей в зависимости от их стоимости')
plt.xlabel('Стоимость автомобиля')
plt.ylabel('Количество машин')
plt.grid(True)
plt.show()
```

Мини-вывод по таблице:
наибольшее кол-во автомобилей представлено в ценовом диапазоне от 5000 до 15000, идет значительное снижение кол-ва авто при цене свыше 20000 -> преобладает низкобюджетный сегмент авторынка



```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x='price', y='mileage', hue='tax', data=dataframe, palette='coolwarm')
plt.title('Влияния пробега на стоимость')
plt.xlabel('Стоимость')
plt.ylabel('Пробег')
plt.grid(True)
plt.legend(title='Налог')
plt.show()
```

Мини-вывод по таблице: чем выше пробег автомобиля, тем, как правило, ниже его стоимость

```python
avg_price = dataframe.groupby('model')['price'].mean()

plt.figure(figsize=(25, 10))
avg_price.plot(kind='bar', color='pink', edgecolor='black')
plt.title('Влияние модели на цену')
plt.xlabel('Модель')
plt.ylabel('Стоимость')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
```

Мини-вывод по таблице: самая дорогая модель автомобилей Ford - mustang, а самая дешевая = streetka


```python
plt.figure(figsize=(25, 10))
sns.boxplot(x='enginesize', y='price', data=dataframe, palette='Set2')
plt.title('Распределение стоимости в завимости от двигателя')
plt.xlabel('Объем двигателя')
plt.ylabel('Стоимость')
plt.grid(True)
plt.show()
```

Мини-вывод по таблице: с увеличеним объема двигателя растет и цена автомобиля, более компактные двигатели, такие как 1.0 и 1.1, соответствуют более низким ценам, тогда как более крупные двигатели, например, 3.2 и 5.0, существенно дороже


```python
plt.figure(figsize=(35, 10))
sns.scatterplot(x='price', y='model', hue='transmission', data=dataframe, palette='coolwarm')
plt.title('Зависимость стоимости от коробчки передач и модели авто')
plt.xlabel('Стоимость')
plt.ylabel('Модель авто')
plt.grid(True)
plt.legend(title='Коробка передач')
plt.show()
```

Мини-вывод по таблице: mustang почти не имеет механической коробки передач и от этого, в том числе, на него более высокие цены; механическая коробка распространена на ранее выпущенных моделях и имеет меньшую стоимость
